```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv(r"D:\ML\Scikit-Learn-Tutorial-\New ML\train (1).csv")[['Age','Pclass','Fare','Survived']]
df.head()
```

```python
df.isnull().mean()*100
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y , test_size=0.02, random_state=42)
```

```python
knnimput = KNNImputer(n_neighbors=5, weights='distance')

X_train_trn = knnimput.fit_transform(X_train)
X_test_trn = knnimput.transform(X_test)
```

```python
lr = LogisticRegression()
lr.fit(X_train_trn,y_train)

y_pred = lr.predict(X_test_trn)
print(accuracy_score(y_test, y_pred))
```

```python
si = SimpleImputer()
X_train_si = si.fit_transform(X_train)
X_test_si = si.transform(X_test)
```

```python
lr = LogisticRegression()
lr.fit(X_train_si, y_train)
y_pred = lr.predict(X_test_si)
print(accuracy_score(y_test, y_pred))
```
